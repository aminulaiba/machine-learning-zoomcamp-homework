##### Question No 7

```python
df_asia = df[df['origin'] == 'Asia']
df_subset = df_asia[['vehicle_weight', 'model_year']].head(7)
df_subset
```

```python
X = df_subset.values
X
```

```python
XTX = X.T @ X
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
w = XTX_inv @ X.T @ y
w.sum()
```
