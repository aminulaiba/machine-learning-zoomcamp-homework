### Q3:  Conversion probability (loaded model)

```python
import pickle
```

```python
model_file = 'pipeline.bin'
```

```python
with open(model_file, 'rb') as f_in:
    dv, model = pickle.load(f_in)
```

```python
dv, model
```

```python
customar = {
  "lead_source": "paid_ads",
  "industry": "technology",
  "employment_status": "employed",
  "location": "north_america",
  "number_of_courses_viewed": 2,
  "annual_income": 79276.0,
  "interaction_count": 4,
  "lead_score": 0.41
}
```

```python
x = dv.transform([customar])
x
```

```python
x.toarray()
```

```python
model.predict(x)
```

```python
model.predict_proba(x)
```

```python
round(model.predict_proba(x)[0, 1], 3)
```

### Q4: Conversion probability (API)

```python
import requests
```

```python
url = 'http://localhost:9696/predict'
```

```python
new_customar = {
  "lead_source": "organic_search",
  "industry": "technology",
  "employment_status": "employed",
  "location": "europe",
  "number_of_courses_viewed": 4,
  "annual_income": 80304.0,
  "interaction_count": 7,
  "lead_score": 0.74
}
new_customar
```

```python
respons = requests.post(url, json=new_customar)
respons
```

```python
respons.json()
```

```python
round(respons.json()['conversion_probability'], 3)
```
